# Как найти оптимальный размер групп при AB тесте 

Оптимальным считаем разбиение, которое минимизирует дисперсию статистики 

Почему? Так как чем меньше дисперсия, тем меньше хвосты распределения, то есть доверительный интервал будет меньше 

Иначе можно сказать, что мы снижаем MDE, так как $MDE = (z_{1 - \alpha} + z_{1 - \beta} \cdot \sqrt{\frac{\sigma_{X}^2}{n_X} + \frac{\sigma_{Y}^2}{n_Y}})$

Задача оптимизации (одна контрольная и $k$ тестовых групп по $m$ человек в каждой): 

\begin{cases} 
\frac{1}{n} + \frac{1}{m} \to \min_{n, m} \\ 
n + k * m = N 
\end{cases}

$$m = \frac{N}{k + \sqrt{k}}, \quad n = \frac{N}{\sqrt{k} + 1}$$

Если k = 1, то есть одна тестовая группа, получим 

$$m = \frac{N}{2}, \quad n = \frac{N}{2}$$
